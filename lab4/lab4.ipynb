```python
import cv2
img= cv2.imread('space.jpg')

threshold_value=[0,50,100,150,200]

for threshold_value in threshold_value:
    ret,thresh = cv2.threshold(img, threshold_value , 255 , cv2.THRESH_BINARY)
    cv2.imshow(f'threshold value ({threshold_value})', thresh)

cv2.imshow('original image', img)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python

import matplotlib.pyplot as plt
import numpy as np

from skimage import data ,img_as_float
from skimage import exposure

img = data.moon()

p2,p98 = np.percentile(img, (2,98))
img_rescale = exposure.rescale_intensity(img, in_range=(p2,p98))

fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.title('original image')
fig.add_subplot(2,1,2)
plt.hist(img.flat,bins=256,range=(0,255))
plt.xlabel('gray level values')
plt.ylabel('number of pixels')
plt.show()

fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.title('contrast stretch')
fig.add_subplot(2,1,2)
plt.hist(img_rescale.flat,bins=256,range=(0,255))
plt.xlabel('gray level values')
plt.ylabel('number of pixels')
plt.show()
```

```python
#task1
img = data.moon()

p3,p80 = np.percentile(img, (3,80))
img_rescale = exposure.rescale_intensity(img, in_range=(p3,p80))

fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.title('rescale image with 3 and 80 percentiles')
fig.add_subplot(2,1,2)
plt.hist(img_rescale.flat,bins=256,range=(0,255))
plt.xlabel('gray level values')
plt.ylabel('number of pixels')
plt.show()
```

```python
#task2
moonEQ=exposure.equalize_hist(img)

fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(moonEQ,cmap='gray')
plt.axis('off')
plt.title('histogram equalized image')
fig.add_subplot(2,1,2)
plt.hist(moonEQ.flat,bins=256,range=(0,1))
plt.xlabel('gray level values')
plt.ylabel('number of pixels')
plt.show()
```

```python
#task3
from skimage.color import rgb2gray
rocket=data.rocket()
chelsea=data.chelsea()

roc_gray=rgb2gray(rocket)
chel_gray=rgb2gray(chelsea)

matched = exposure.match_histograms(roc_gray,chel_gray)


fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(roc_gray,cmap='gray')
plt.axis('off')
plt.title('rocket')

fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(chel_gray,cmap='gray')
plt.axis('off')
plt.title('chelsea')

fig = plt.figure(figsize=(10,7))
fig.add_subplot(2,1,1)
plt.imshow(matched,cmap='gray')
plt.axis('off')
plt.title('matched histogram')
```
